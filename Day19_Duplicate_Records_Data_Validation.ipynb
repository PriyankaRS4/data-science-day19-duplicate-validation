# Task 19 — Duplicate Records and Data Validation

Detect duplicate rows and perform simple validation checks to identify inconsistent or invalid records.

```python
import pandas as pd
df = pd.read_csv('sales_data.csv')
df.shape
```

```python
# Duplicate record analysis
print('Duplicate rows:', df.duplicated().sum())
duplicate_records = df[df.duplicated(keep=False)]
duplicate_records
```

```python
# Clean dataset without unwanted duplicates
cleaned_df = df.drop_duplicates().copy()
cleaned_df.to_csv('sales_data_cleaned.csv', index=False)
print('Before:', df.shape)
print('After :', cleaned_df.shape)
```

```python
# Validation 1: missing values
print(df.isna().sum())
```

```python
# Validation 2: duplicate rows
print('Duplicate rows:', df.duplicated().sum())
```

```python
# Validation 3: Sales_Amount should not be negative
print('Negative Sales_Amount:', (df['Sales_Amount'] < 0).sum())
```

```python
# Validation 4: Quantity_Sold should be positive
print('Non-positive Quantity_Sold:', (df['Quantity_Sold'] <= 0).sum())
```

```python
# Validation 5: Discount should be between 0 and 1
print('Discount outside 0-1:', ((df['Discount'] < 0) | (df['Discount'] > 1)).sum())
```

```python
# Validation 6: expected categories
for col in ['Region','Customer_Type','Sales_Channel']:
    print(col, sorted(df[col].dropna().unique()))
```

```python
# Validation 7: combined field consistency
expected = df['Region'].astype(str) + '-' + df['Sales_Rep'].astype(str)
print('Inconsistent Region_and_Sales_Rep:', (df['Region_and_Sales_Rep'].astype(str) != expected).sum())
```

## Conclusion

The sales dataset was checked for duplicates and common data-quality issues. A cleaned dataset was created using `drop_duplicates()`, and multiple validation checks were performed before analysis.
